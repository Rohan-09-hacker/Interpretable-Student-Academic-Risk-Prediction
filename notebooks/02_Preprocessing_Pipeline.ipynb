## 📌 Pipeline Objectives
1. Define the ground-truth binary label: **`At_Risk`** ($G3 < 10$).
2. Engineer domain-specific indicators (Parent education, Study-to-free-time ratio, Alcohol consumption index).
3. Mitigate data leakage by excluding downstream grade columns ($G1, G2, G3$) for true early prediction.
4. Apply **StandardScaler** to numerical columns and **One-Hot Encoding** to categorical columns.
5. Perform stratified train-test splitting (80/20) and export processed datasets to `data/processed/`.

```python
import sys
import os
import pandas as pd
import numpy as np

# Add src/ to path
sys.path.append(os.path.abspath(os.path.join("..", "src")))
from data_preprocessing import load_raw_data, engineer_features, preprocess_and_split

print("Preprocessing module loaded successfully!")
```

## 1. Inspect Raw Data & Engineered Features

```python
raw_path = os.path.join("..", "data", "student-mat.csv")
raw_df = load_raw_data(raw_path)
print(f"Raw dataset shape: {raw_df.shape}")

featured_df = engineer_features(raw_df)
print(f"Featured dataset shape: {featured_df.shape}")
featured_df[['studytime', 'freetime', 'study_freetime_ratio', 'Medu', 'Fedu', 'parent_edu_avg', 'alcohol_index']].head()
```
